# Knowledge Decay Prediction using Machine Learning

## Project Overview

Knowledge decay is the gradual loss of previously learned information when knowledge is not regularly reinforced.

This project presents a machine learning pipeline that predicts whether a learner is likely to experience a future forgetting event based on historical learning interactions.

The complete workflow includes:

- Data preprocessing
- Feature engineering
- Mastery detection
- Forget event construction
- Machine learning model development
- Model evaluation
- Prediction pipeline
- Demonstration dashboard

The proposed system can support adaptive learning platforms by identifying learners who may require timely revision.

📒 Section 2 — Problem Statement

# Problem Statement

Many online learning platforms evaluate students only after forgetting has already occurred.

There is limited support for predicting future forgetting using historical learning behaviour.

The objective of this project is to develop a machine learning model capable of identifying learners at risk of forgetting before significant knowledge decay occurs.

📒 Section 7 — Project Limitations

# Project Limitations

The current implementation has several limitations:

- The study was conducted using a single educational dataset.
- Only historical interaction features were considered.
- Cold-start learners require an initial interaction history before prediction.
- The evaluation used a stratified train-test split rather than a strictly chronological split.
- The system currently operates as an offline prediction pipeline.
